# REST API demo: поиск фильма через OMDb API

В этом ноутбуке пользователь вводит название фильма и API-ключ OMDb, а Python отправляет REST-запрос и разбирает JSON-ответ.

## 1. Подключаем библиотеки

`requests` нужен для HTTP-запроса, `json` - для красивого вывода JSON, `getpass` - для безопасного ввода API-ключа.

```python
import json
import getpass

import requests
```

## 2. Вводим фильм и API-ключ

API-ключ вводится вручную и не сохраняется в файлах проекта.

```python
title = input("Введите фильм для поиска: ").strip()
api_key = getpass.getpass("Введите ваш OMDb API key: ").strip()

if not title:
    raise ValueError("Название фильма не может быть пустым")

if not api_key:
    raise ValueError("API-ключ не может быть пустым")
```

## 3. Отправляем GET-запрос

Параметр `t` содержит название фильма, а `apikey` - ключ доступа к OMDb API.

```python
response = requests.get(
    "http://www.omdbapi.com/",
    params={
        "t": title,
        "apikey": api_key,
    },
    timeout=10,
)
response.raise_for_status()
```

## 4. Превращаем ответ в JSON

Сначала выводим полный JSON, чтобы увидеть все поля, которые вернул сервис.

```python
json_data = response.json()
print(json.dumps(json_data, ensure_ascii=False, indent=2))
```

## 5. Достаём нужные поля

Используем `.get()`, чтобы безопасно получить значения из словаря Python.

```python
if json_data.get("Response") == "False":
    print("Фильм не найден:", json_data.get("Error", "неизвестная ошибка"))
else:
    movie_title = json_data.get("Title", "Нет данных")
    year = json_data.get("Year", "Нет данных")
    genre = json_data.get("Genre", "Нет данных")
    director = json_data.get("Director", "Нет данных")
    imdb_rating = json_data.get("imdbRating", "Нет данных")

    print(f"Название: {movie_title}")
    print(f"Год: {year}")
    print(f"Жанр: {genre}")
    print(f"Режиссёр: {director}")
    print(f"Рейтинг IMDb: {imdb_rating}")
```
