# REST API demo: OMDb API и `.env`

В этой ветке API-ключ берётся из файла `.env`, а не вводится с клавиатуры. Ветка `env-feature` не вливается в `master`, потому что показывает отдельный вариант работы с секретами.

## 1. Подключаем библиотеки

`python-dotenv` загружает переменные окружения из `.env`.

```python
import json
import os

import requests
from dotenv import load_dotenv
```

## 2. Загружаем ключ из `.env`

Создайте файл `.env` по примеру `.env.example`: `OMDB_API_KEY=ваш_ключ`.

```python
load_dotenv()

title = input("Введите фильм для поиска: ").strip()
api_key = os.getenv("OMDB_API_KEY")

if not title:
    raise ValueError("Название фильма не может быть пустым")

if not api_key:
    raise ValueError("Ключ OMDB_API_KEY не найден в .env")
```

## 3. Отправляем запрос к OMDb API

```python
response = requests.get(
    "http://www.omdbapi.com/",
    params={
        "t": title,
        "apikey": api_key,
    },
    timeout=10,
)
response.raise_for_status()
```

json_data = response.json()
